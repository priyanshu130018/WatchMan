### Fetch All Trending Movies and Create Dataframe

```python
import requests
import pandas as pd
import time
```

```python
api_key = ""

session = requests.Session()

dfs = []

for i in range(1, 429):

    url = "https://api.themoviedb.org/3/movie/top_rated"

    params = {
        "api_key": api_key,
        "language": "en-US",
        "page": i
    }

    try:
        response = session.get(
            url,
            params=params,
            timeout=30
        )

        print("Page:", i, "Status:", response.status_code)

        response.raise_for_status()

        temp_df = pd.DataFrame(response.json()["results"])[
            [
                "id",
                "title",
                "overview",
                "release_date",
                "popularity",
                "vote_average",
                "vote_count"
            ]
        ]

        dfs.append(temp_df)

        time.sleep(1)

    except requests.exceptions.ConnectionError as e:
        print("Connection error on page:", i)
        print(e)
        time.sleep(5)
        continue

    except requests.exceptions.RequestException as e:
        print("Request error on page:", i)
        print(e)
        break

df = pd.concat(dfs, ignore_index=True)

df.shape
```
